```python
img = cv2.imread(sample_dir)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(img_gray)
```

```python
img_small = cv2.resize(img, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)

img_resized = cv2.resize(img, (200, 200), interpolation=cv2.INTER_AREA)

cv2_imshow(img_small)
cv2_imshow(img_resized)
```

```python
img = np.zeros((400,400,3), dtype = np.uint8)

cv2.line(img, (50,50), (350,50), (0,255,0),3)
cv2_imshow(img)
```

```python
img_rect = img.copy()

cv2.rectangle(img_rect, (100,100), (300,200), (255,0,0), 2)

cv2.rectangle(img_rect, (100,250), (300,350), (0,0,255), -1)

cv2_imshow(img_rect)
```

```python
img_circle = img.copy()

cv2.circle(img_circle, (200,200), 40, (0,255,255), 4)

cv2.circle(img_circle, (200,200), 80, (255,0,255), 3)
cv2_imshow(img_circle)
```

```python
img_elipse= img.copy()

cv2.ellipse(img_elipse, (200,200),(100,50), 30, 0, 360, (255,255,0), 3)
cv2_imshow(img_elipse)
```

```python
img_poly = img.copy()
pts = np.array([[50,300],[200,150],[350,300]],np.int32)
pts = pts.reshape((-1,1,2))
cv2.polylines(img_poly, [pts], isClosed=True, color=(0,255,255),thickness=3)
cv2_imshow(img_poly)
```

```python
img_text = img.copy()
cv2.putText(img_text,"Hello World", (50,200), cv2.FONT_HERSHEY_SIMPLEX, 1.5, (255,255,255), 3, cv2.LINE_AA)
cv2_imshow(img_text)
```

```python
# Ahorro de linea
img_bgr = cv2.imread(sample_dir)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_bgr[...,::-1])
```

```python
b, g, r = cv2.split(img_bgr)

zeros = np.zeros_like(b)

img_b = cv2.merge([b, zeros, zeros])
img_g = cv2.merge([zeros, g, zeros])
img_r = cv2.merge([zeros, zeros, r])

cv2_imshow(img_b)
cv2_imshow(img_g)
cv2_imshow(img_r)
```

```python
img_bc = cv2.convertScaleAbs(img_bgr, alpha=1.5, beta=50)
cv2_imshow(img_bc)
```

```python
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
```

```python
img_eq = cv2.equalizeHist(img_gray)
cv2_imshow(img_eq)
```

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)

h,s,v = cv2.split(img_hsv)

cv2_imshow(h)
cv2_imshow(s)
cv2_imshow(v)
```

```python
loower_red1 = np.array([0,120,70])
upper_red1 = np.array([10,255,255])
mask1 = cv2.inRange(img_hsv, loower_red1, upper_red1)
cv2_imshow(mask1)
```

```python
img_hsv2 = img_hsv.copy()
img_hsv2[:,:,1] = np.clip(img_hsv2[:,:,1].astype('float32')*1.2,0,255).astype('uint8')
img_stat_boost = cv2.cvtColor(img_hsv2, cv2.COLOR_HSV2BGR)
cv2_imshow(img_stat_boost)
```

```python
img_lab = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2LAB)

l,a,b = cv2.split(img_lab)

cv2_imshow(l)
cv2_imshow(a)
cv2_imshow(b)
```

```python
clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8,8))
clahe = clahe.apply(l)
img_lab_clahe = cv2.merge([clahe,a,b])
img_clahe = cv2.cvtColor(img_lab_clahe, cv2.COLOR_LAB2BGR)
cv2_imshow(img_clahe)
```

```python
img_gray = cv2.imread(sample_dir, cv2.IMREAD_GRAYSCALE)

_, th_bin = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

cv2_imshow(th_bin)
cv2_imshow(th_bin)
```

```python
th_adapt = cv2.adaptiveThreshold(img_gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2)
cv2_imshow(th_adapt)
```

```python
img_bgr = cv2.imread(sample_dir)
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)

lower_blue = (100,150,0)
upper_blue = (140,255,255)
mask_blue = cv2.inRange(img_hsv, lower_blue, upper_blue)
res_blue = cv2.bitwise_and(img_bgr, img_bgr, mask=mask_blue)

cv2_imshow(mask_blue)
cv2_imshow(res_blue)
```

```python
edges = cv2.Canny(img_gray, 100, 200)
cv2_imshow(img_gray)
cv2_imshow(edges)
```

```python
_, th_bin = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

contours, _ = cv2.findContours(th_bin, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
imag_contours = cv2.cvtColor(th_bin, cv2.COLOR_GRAY2BGR)

cv2.drawContours(imag_contours, contours, -1, (0,255,0), 2)

cv2_imshow(imag_contours)
```
